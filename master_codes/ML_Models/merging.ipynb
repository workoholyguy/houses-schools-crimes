```python
import pandas as pd

crime_df = pd.read_csv("crime2024_with_zipcodes.csv")
school_df = pd.read_csv("school_quality_ratings.csv")
zip_df = pd.read_csv("zip_neighborhoods.csv")

# Rename ZIP columns if needed
crime_df = crime_df.rename(columns={'zipcode': 'ZIP Code'})

school_zip_merged = pd.merge(school_df, zip_df, on='ZIP Code', how='inner')

final_merged_df = pd.merge(school_zip_merged, crime_df, on='ZIP Code', how='left')
```
